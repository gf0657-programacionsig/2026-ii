## Variables

Una **variable** es un nombre asignado a un valor en la memoria de la computadora, como se vio en los algoritmos y programas de la [lección de introducción a la programación](../i-introduccion-ciencia-datos-programacion/06-introduccion-programacion.md). El valor se asigna mediante el operador `=`.

````python
# Asignación de valores a dos variables
poblacion = 352381  # población del cantón de San José (INEC, Censo 2022)
area = 44.6         # área del cantón en km2
````

````python
# Impresión del valor de una variable
print(poblacion)
````

````python
# En un cuaderno de notas, el valor también se despliega
# al escribir la variable al final de una celda de código
area
````

````python
# El valor de una variable puede actualizarse
# (aquí, con el resultado de una operación sobre su valor anterior)
poblacion = poblacion + 1000

poblacion
````

````python
# Las variables también almacenan hileras
canton = "San José"

canton
````

````python
# type() también funciona con variables
type(canton)
````

Los nombres de las variables deben cumplir varias condiciones:

- No pueden empezar con un número (sí pueden usar números en el resto del nombre).
- No pueden contener espacios ni la mayoría de los símbolos (ej. `-`, `#`, `%`); para separar palabras se usa el guion bajo (`_`).
- No pueden ser [palabras reservadas de Python](https://docs.python.org/es/3/reference/lexical_analysis.html#keywords) (ej. `if`, `else`, `True`).
- Distinguen entre mayúsculas y minúsculas: `area`, `Area` y `AREA` son tres variables diferentes.

Además, la [guía de estilo PEP 8](https://peps.python.org/pep-0008/#function-and-variable-names) recomienda usar nombres en minúscula, con guiones bajos, y este curso pide que sean **significativos**: `poblacion` y `area` dicen más que `p` y `a`, en línea con la legibilidad estudiada en la [lección anterior](08-introduccion-python.md). Note también que en los nombres se evitan las tildes y la eñe (`poblacion`, no `población`): aunque Python las acepta, su uso puede causar problemas con codificaciones de caracteres y teclados distintos.

## Expresiones y sentencias

Una **sentencia** (*statement*) es una instrucción completa que el interpretador puede ejecutar, como una asignación (`area = 44.6`) o un llamado a `print()`. Un programa es una secuencia de sentencias (Severance, 2016).

Una [**expresión**](https://docs.python.org/es/3/reference/expressions.html) es una combinación de:

- [Literales](https://docs.python.org/es/3/reference/lexical_analysis.html#literals): valores constantes de números, hileras, booleanos y otros tipos (ej. `44.6`, `"Costa Rica"`, `True`).
- Variables.
- Los dos anteriores combinados mediante [operadores](https://docs.python.org/es/3/reference/lexical_analysis.html#operators) o funciones.

Toda expresión produce un valor, que puede ser numérico, textual, booleano o de otro tipo.

## Condicionales

### La sentencia `if`

Los **condicionales** ejecutan un bloque de sentencias solo si se cumple una condición. En Python se implementan con la sentencia [if](https://docs.python.org/es/3/reference/compound_stmts.html#the-if-statement), cuya forma más básica es:

```python
if <condición>:
    <bloque de sentencias>
```

`<condición>` es una expresión booleana. Si es verdadera, se ejecuta el bloque que está después de los dos puntos (`:`), formado por las sentencias indentadas (i.e. tabuladas) con la misma cantidad de espacios. Por ejemplo, el siguiente programa evalúa la magnitud de un sismo:

````python
magnitud = 6.2

print(f"Magnitud del sismo: {magnitud}")

if magnitud >= 6.0:
    print("El sismo es FUERTE.")
    print("Puede causar daños en estructuras vulnerables.")

print("Fin del programa.")
````

Después de ejecutar el bloque (si la condición se cumple), el programa continúa con las sentencias que están fuera de él: la última línea se imprime siempre, sin importar la magnitud. Pruebe cambiar el valor de `magnitud` y vuelva a ejecutar la celda.

### La cláusula `else`

La cláusula opcional `else` especifica un bloque alterno, que se ejecuta cuando la condición del `if` no se cumple:

```python
if <condición>:
    <bloque de sentencias>
else:
    <bloque de sentencias>
```

Cada sentencia `if` puede tener solo una cláusula `else`. Por ejemplo:

````python
magnitud = 4.5

print(f"Magnitud del sismo: {magnitud}")

if magnitud >= 6.0:
    print("El sismo es FUERTE.")
else:
    print("El sismo NO es fuerte.")

print("Fin del programa.")
````

### La cláusula `elif`

La cláusula opcional `elif` ("else if") permite evaluar condiciones adicionales entre el `if` y el `else`:

```python
if <condición>:
    <bloque de sentencias>
elif <condición>:
    <bloque de sentencias>
else:
    <bloque de sentencias>
```

Puede usarse cualquier cantidad de cláusulas `elif`, con o sin `else` final. Las condiciones se evalúan en orden y solo se ejecuta el bloque de la primera que resulte verdadera. Por ejemplo, el siguiente programa clasifica la magnitud de un sismo en tres categorías (clasificación simplificada, con fines didácticos):

````python
magnitud = 4.5

print(f"Magnitud del sismo: {magnitud}")

if magnitud < 4.0:
    print("El sismo es LEVE: usualmente no causa daños.")
elif magnitud < 6.0:
    print("El sismo es MODERADO: puede causar daños menores.")
else:
    print("El sismo es FUERTE: puede causar daños en estructuras vulnerables.")

print("Fin del programa.")
````

### Bloques e indentación

Un **bloque** es un conjunto de sentencias contiguas indentadas con la misma cantidad de espacios. La indentación no es decorativa: es la forma en la que Python delimita los bloques, por lo que un espacio de más o de menos cambia el significado del programa o produce un error. La [guía de estilo PEP 8](https://peps.python.org/pep-0008/#indentation) recomienda indentar con **cuatro espacios** y no mezclar espacios con tabuladores; los editores del curso (Colab y VS Code) insertan los espacios automáticamente al escribir los dos puntos y presionar `Enter`.

## Excepciones

Las [**excepciones**](https://docs.python.org/es/3/tutorial/errors.html) son el mecanismo de Python para manejar los errores que se producen durante la ejecución de un programa: cuando ocurre uno, el curso normal del programa se interrumpe y se "levanta" (*raise*) una excepción, que puede ser atendida por otro bloque de código.

Por ejemplo, la función `float()` produce un error si la hilera que recibe no corresponde a un número. Es un caso frecuente con datos de Costa Rica y de otros países donde el separador decimal es la coma, que `float()` no acepta:

````python
# Esta celda produce un error (ValueError) de forma intencional
float("44,6")
````

### Las sentencias `try` y `except`

Las sentencias [try y except](https://docs.python.org/es/3/reference/compound_stmts.html#the-try-statement) implementan el manejo de excepciones: en el bloque `try` se coloca el código que puede producir el error y en el bloque `except`, el código que lo maneja.

```python
try:
    <bloque de sentencias que puede producir un error>
except:
    <bloque de sentencias que maneja el error>
```

El siguiente ejemplo maneja el error anterior y lo sustituye por un mensaje más claro:

````python
dato = "44,6"

try:
    area = float(dato)
    print(f"Área: {area} km2")
except:
    print(f"El valor '{dato}' no es un número válido: use punto decimal, no coma.")
````

El manejo de excepciones es especialmente útil al leer datos digitados por una persona. La función [input()](https://docs.python.org/es/3/library/functions.html#input) lee una hilera desde el teclado; como este cuaderno se publica ya ejecutado, el siguiente programa se muestra sin ejecutar — cópielo en una celda de código y pruébelo en Colab o en VS Code:

```python
altitud_hilera = input("Ingrese la altitud del lugar en metros: ")

try:
    altitud = float(altitud_hilera)
    print(f"La altitud ingresada es {altitud} m.")
except:
    print("Por favor ingrese un número (con punto decimal).")
```

## Resumen

- Todo valor de Python tiene un **tipo de datos**; los básicos son `int`, `float`, `bool` y `str`, la función `type()` los identifica y las funciones `int()`, `float()` y `str()` convierten entre ellos.
- Una **variable** es un nombre asignado a un valor con el operador `=`; los nombres deben ser válidos y significativos, y los **comentarios** (`#`) documentan el código.
- Un programa es una secuencia de **sentencias**; las **expresiones** combinan literales, variables, operadores y funciones para producir un valor, que puede ser aritmético o **booleano** (`True`/`False`). Las **f-strings** presentan resultados con formato.
- Los **condicionales** (`if`, `elif`, `else`) ejecutan bloques según se cumplan expresiones booleanas; los **bloques** se delimitan con la indentación (cuatro espacios).
- Las **excepciones** (`try`/`except`) manejan los errores de ejecución, como los de conversión de datos.
