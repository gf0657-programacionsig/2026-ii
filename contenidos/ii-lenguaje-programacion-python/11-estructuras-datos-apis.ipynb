## Objetos y métodos

En Python, todo valor es un **objeto**: además de su valor, tiene asociado un conjunto de operaciones propias de su tipo, llamadas **métodos**. Los métodos se invocan con la **notación de punto**: `objeto.método()`. A diferencia de una función como `len()`, que se invoca por sí sola, un método se invoca *sobre* un objeto:

````python
nombre_pais = "Costa Rica"

# len() es una función: recibe el objeto como argumento
print(len(nombre_pais))

# upper() es un método de las hileras: se invoca sobre el objeto
print(nombre_pais.upper())
````

Esta notación ya apareció con los módulos (`math.sqrt()`) y será omnipresente en el resto del curso: `df.head()`, `mapa.add_to()` y similares son métodos de objetos de pandas y folium. Cada tipo de datos tiene sus propios métodos, como se verá en las secciones siguientes.

## Hileras

Una hilera (`str`) es una **secuencia** de caracteres, y como toda secuencia puede indexarse y recortarse:

````python
# Indexación: el primer carácter está en la posición 0
# y los índices negativos cuentan desde el final
print(nombre_pais[0])
print(nombre_pais[-1])
````

````python
# Rebanado (slicing): [inicio:fin] extrae desde inicio hasta fin - 1
nombre_pais[0:5]
````

Las hileras ofrecen numerosos [métodos](https://docs.python.org/es/3/library/stdtypes.html#string-methods). Algunos de los más usados en la limpieza de datos —tarea central de las próximas unidades— son:

````python
# upper(), lower(): mayúsculas y minúsculas
print("puntarenas".upper())
print("LIMÓN".lower())
````

````python
# strip(): elimina los espacios al inicio y al final
# (frecuentes en datos digitados a mano)
"  San José  ".strip()
````

````python
# replace(): reemplaza una subhilera por otra
"provincia de Cartago".replace("provincia", "ciudad")
````

````python
# split(): divide la hilera y retorna una LISTA
"San José, Alajuela, Cartago".split(", ")
````

Los métodos de hileras **retornan copias**: la hilera original no cambia (las hileras son **inmutables**). Por último, el operador `in` verifica si una subhilera está contenida en otra:

````python
"Rica" in nombre_pais
````

## Listas

Las listas (`list`), usadas desde la parte II, son secuencias **mutables**: pueden modificarse después de creadas. Igual que las hileras, se indexan y rebanan; además, tienen métodos que las modifican:

````python
cantones = ["San José", "Escazú", "Desamparados"]

# append(): agrega un elemento al final
cantones.append("Aserrí")
cantones
````

````python
# remove(): elimina la primera ocurrencia de un valor
cantones.remove("Escazú")
cantones
````

````python
# sort(): ordena la lista EN EL LUGAR (la modifica y no retorna nada)
cantones.sort()
cantones
````

````python
# sorted() es la alternativa como función: retorna una COPIA ordenada
# sin modificar la original
numeros = [3, 1, 2]
print(sorted(numeros))
print(numeros)
````

````python
# El operador in también funciona con listas
"Aserrí" in cantones
````

La diferencia entre `sort()` (método que modifica y retorna `None`) y `sorted()` (función que retorna una copia) es una fuente clásica de confusión, emparentada con la distinción retornar/imprimir de la parte II; el ejercicio 2 la pone a prueba.

## Tuplas

Las tuplas (`tuple`) son secuencias **inmutables**: una vez creadas, no pueden modificarse. Se escriben entre paréntesis y son idóneas para datos que van juntos y no deben cambiar, como un par de coordenadas:

````python
# Una tupla con la latitud y la longitud aproximadas de la ciudad de San José
punto_san_jose = (9.93, -84.08)

# Se indexa como cualquier secuencia
punto_san_jose[0]
````

````python
# La inmutabilidad protege los datos: intentar modificar una tupla
# produce un TypeError (error intencional)
punto_san_jose[0] = 10.0
````

````python
# Desempaquetado: asignar los elementos de una tupla a variables individuales
latitud, longitud = punto_san_jose

print(f"Latitud: {latitud}, longitud: {longitud}")
````

## Diccionarios

Los diccionarios (`dict`) almacenan pares **clave: valor** y son la estructura más usada para representar entidades con atributos — un cantón, una provincia, un registro de una especie. Se escriben entre llaves:

````python
# Un diccionario con los datos de una provincia
# (población: INEC, Censo 2022; área: IGN)
provincia = {
    "nombre": "Guanacaste",
    "poblacion": 412808,
    "area": 10140.7
}

# Acceso al valor de una clave
provincia["poblacion"]
````

````python
# Agregar (o modificar) una clave es una asignación
provincia["capital"] = "Liberia"
provincia
````

````python
# El operador in verifica la EXISTENCIA de una clave
print("capital" in provincia)
print("densidad" in provincia)
````

````python
# get() retorna el valor de una clave, o un valor por defecto si no existe
# (a diferencia de provincia["densidad"], que produciría un error)
provincia.get("densidad", "sin calcular")
````

````python
# items() retorna los pares (clave, valor), útiles para iterar
# con un for y desempaquetado de tuplas
for clave, valor in provincia.items():
    print(f"{clave}: {valor}")
````

### Listas de diccionarios

La combinación más importante para este curso es la **lista de diccionarios**: cada elemento es una entidad completa, con sus atributos nombrados. Compárese con las tres listas paralelas de la [parte II](https://gf0657-programacionsig.github.io/2026-ii/funciones-iteracion/) — aquí los datos de cada provincia viajan juntos y el código se lee solo:

````python
# Las mismas provincias de la parte II, ahora como lista de diccionarios
# (población: INEC, Censo 2022; área: IGN)
provincias = [
    {"nombre": "San José", "poblacion": 1601167, "area": 4965.9},
    {"nombre": "Alajuela", "poblacion": 1035466, "area": 9757.5},
    {"nombre": "Cartago", "poblacion": 545092, "area": 3124.7},
    {"nombre": "Heredia", "poblacion": 479117, "area": 2657.0},
    {"nombre": "Guanacaste", "poblacion": 412808, "area": 10140.7},
    {"nombre": "Puntarenas", "poblacion": 500166, "area": 11265.7},
    {"nombre": "Limón", "poblacion": 470383, "area": 9188.5}
]

# Densidad de cada provincia, sin índices ni listas paralelas
for provincia in provincias:
    densidad = provincia["poblacion"] / provincia["area"]
    print(f"{provincia['nombre']}: {densidad:.1f} hab/km2")
````

Esta estructura —una lista de "registros", cada uno con los mismos atributos— es exactamente la forma de una tabla de datos: en la unidad III, pandas la convertirá en su estructura central, el *DataFrame*.

## Servicios web y APIs REST

Las estructuras de datos recién estudiadas son la llave de acceso a los **servicios web**: programas que entregan datos a otros programas a través de Internet. Una [**API**](https://es.wikipedia.org/wiki/API) (*Application Programming Interface*, interfaz de programación de aplicaciones) define cómo solicitar esos datos; las de tipo [**REST**](https://es.wikipedia.org/wiki/Transferencia_de_Estado_Representacional), las más comunes, funcionan sobre el mismo protocolo [HTTP](https://es.wikipedia.org/wiki/Protocolo_de_transferencia_de_hipertexto) de la Web:

- El programa **cliente** envía una solicitud, típicamente con el método **GET** (pedir datos), a una **URL** que identifica el recurso, con **parámetros** que precisan la consulta.
- El **servidor** responde con un **código de estado** —200 significa éxito; 404, recurso no encontrado; ya se han visto en GitHub Pages— y con los datos solicitados.

### JSON

Los datos suelen viajar en formato [**JSON**](https://es.wikipedia.org/wiki/JSON) (*JavaScript Object Notation*): texto estructurado casi idéntico a los diccionarios y las listas de Python. El diccionario de la provincia de Guanacaste, en JSON, luce así:

```json
{
  "nombre": "Guanacaste",
  "poblacion": 412808,
  "area": 10140.7,
  "capital": "Liberia"
}
```

Por eso las estructuras de este cuaderno se estudian junto con las APIs: **consultar una API es recibir diccionarios y listas** (usualmente anidados) y recorrerlos con las herramientas ya conocidas. Este es, además, el formato del archivo `.ipynb` que se examinó en los ejercicios de la [lección de cuadernos](https://gf0657-programacionsig.github.io/2026-ii/cuadernos-jupyter/).

### El paquete requests

El paquete [requests](https://requests.readthedocs.io/) es la biblioteca estándar de facto para hacer solicitudes HTTP en Python. Viene preinstalado en Colab y en el ambiente `geopython`. Su función principal, `requests.get()`, envía una solicitud GET y retorna un objeto de respuesta, cuyo método `json()` convierte los datos JSON en diccionarios y listas de Python:

````python
import requests
````

### Ejemplo: la API de GBIF

[GBIF](https://www.gbif.org/) —la fuente de datos de biodiversidad de la semana 1— ofrece una [API REST pública](https://techdocs.gbif.org/en/openapi/) y sin registro. El siguiente programa consulta el identificador taxonómico (`usageKey`) de la lapa verde (*Ara ambiguus*), la especie del [cuaderno de ejemplo](https://gf0657-programacionsig.github.io/2026-ii/ejemplo-procesos-ciencia-datos/):

````python
# Consulta del identificador taxonómico de una especie
respuesta = requests.get(
    "https://api.gbif.org/v1/species/match",
    params={"name": "Ara ambiguus", "strict": True}
)

# Verificación del código de estado (200 = éxito)
print(respuesta.status_code)
````

````python
# Conversión de la respuesta JSON a estructuras de Python
datos_especie = respuesta.json()

# La respuesta es un diccionario; se extraen valores con las claves
print(datos_especie["scientificName"])
print(datos_especie["usageKey"])
````

Con el identificador, una segunda consulta obtiene los registros de presencia de la especie en Costa Rica. La respuesta anida las estructuras estudiadas: un diccionario cuyo valor `"results"` es una **lista de diccionarios**, uno por registro:

````python
# Consulta de registros de presencia en Costa Rica (los primeros 5 con coordenadas)
respuesta = requests.get(
    "https://api.gbif.org/v1/occurrence/search",
    params={
        "taxonKey": datos_especie["usageKey"],
        "country": "CR",
        "hasCoordinate": True,
        "limit": 5
    }
)

datos_registros = respuesta.json()

# Total de registros disponibles
print(f"Registros de la especie en Costa Rica: {datos_registros['count']}")
````

````python
# Recorrido de la lista de registros: cada uno es un diccionario
for registro in datos_registros["results"]:
    print(
        registro["species"],
        registro.get("year", "sin año"),
        registro["decimalLatitude"],
        registro["decimalLongitude"]
    )
````

Note el uso de `get()` para el año: no todos los registros traen todas las claves, y `get()` evita el `KeyError` con un valor por defecto — el mismo problema de los valores faltantes de la [primera lección](https://gf0657-programacionsig.github.io/2026-ii/introduccion-ciencia-datos/), ahora en código.

Existen APIs públicas para casi cualquier dato geográfico: sismos ([USGS](https://earthquake.usgs.gov/fdsnws/event/1/), usada en los ejercicios), países, clima (algunas, como [OpenWeatherMap](https://openweathermap.org/api), requieren registrarse para obtener una llave de acceso). El patrón es siempre el mismo: `requests.get()` con parámetros, verificar el estado, recorrer las estructuras.

### Un adelanto: los registros en un mapa

Para cerrar la serie de fundamentos, el siguiente código coloca los cinco registros recién descargados en un mapa interactivo con la biblioteca folium, ya instalada en el ambiente del curso. No se estudiará todavía — es un adelanto de la unidad de datos geoespaciales, y la razón de ser de todo lo aprendido en estas tres semanas:

````python
import folium

# Mapa centrado en Costa Rica
mapa = folium.Map(location=[9.97, -84.19], zoom_start=7)

# Un marcador por cada registro de presencia
for registro in datos_registros["results"]:
    folium.Marker(
        location=[registro["decimalLatitude"], registro["decimalLongitude"]],
        popup=registro["species"]
    ).add_to(mapa)

mapa
````

## Resumen

- Todo valor de Python es un **objeto** con **métodos** que se invocan con la notación de punto (`objeto.método()`), la misma de las bibliotecas que vienen (`df.head()`).
- Las **hileras** son secuencias inmutables: se indexan, se rebanan y ofrecen métodos de limpieza (`upper()`, `strip()`, `replace()`, `split()`) que retornan copias.
- Las **listas** son secuencias mutables (`append()`, `remove()`, `sort()`); las **tuplas**, secuencias inmutables idóneas para coordenadas, con **desempaquetado**.
- Los **diccionarios** guardan pares clave: valor; una **lista de diccionarios** representa una tabla de datos y anticipa el *DataFrame* de pandas.
- Una **API REST** entrega datos por HTTP: `requests.get()` con parámetros, código de estado y **JSON**, que se convierte en diccionarios y listas con `.json()`.
- Los **asistentes de IA** se usan desde esta semana para explicar y depurar: prompts con rol, contexto, tarea y formato; verificar las respuestas y declarar el uso.
- Errores típicos de las estructuras: `KeyError` (clave inexistente), `IndexError` (índice fuera de rango) y `AttributeError` (método inexistente).
