## ¿Qué es pandas?

[pandas](https://pandas.pydata.org/) es una biblioteca de Python de código abierto para el **análisis y la manipulación de datos tabulares**. Fue creada por Wes McKinney en 2008, cuando trabajaba en una empresa financiera y necesitaba una herramienta ágil para analizar series de tiempo; su nombre deriva de *panel data*, un término de la econometría. Hoy es la herramienta estándar para trabajar con tablas de datos en Python y la base sobre la que se construyen las bibliotecas geoespaciales de la unidad siguiente: un *GeoDataFrame* de geopandas es un *DataFrame* de pandas con una columna de geometrías.

pandas se apoya en [NumPy](https://numpy.org/), la biblioteca de cómputo numérico de Python, de la que hereda su característica más importante: las operaciones son **vectorizadas**, es decir, se aplican a todos los elementos de una columna a la vez, sin escribir ciclos. Sus dos estructuras de datos son la **Series** (una columna) y el **DataFrame** (una tabla).

Por convención universal, pandas se importa con el alias `pd`:

````python
import pandas as pd

# Versión instalada de pandas
print(pd.__version__)
````

## Series

Una **Series** es un arreglo unidimensional de valores del mismo tipo acompañado de un **índice**: una etiqueta para cada valor. Puede pensarse como una columna de una tabla o como un cruce entre una lista (valores ordenados) y un diccionario (valores con etiqueta). Se crea a partir de una lista:

````python
# Población de las provincias de Costa Rica
# (INEC, Estimación de Población y Vivienda 2022)
poblacion = pd.Series(
    [1601167, 1035464, 545092, 479117, 412808, 500166, 470383],
    index=["San José", "Alajuela", "Cartago", "Heredia", "Guanacaste", "Puntarenas", "Limón"]
)

poblacion
````

A la izquierda se muestra el índice y a la derecha los valores; al final, el tipo de datos (`dtype`) de los valores, en este caso enteros de 64 bits. Si no se especifica un índice, pandas asigna las posiciones 0, 1, 2, ... como en una lista.

Los valores se acceden por su **etiqueta** con `loc[]` o por su **posición** con `iloc[]`:

````python
# Acceso por etiqueta (como en un diccionario)
print(poblacion.loc["Guanacaste"])

# Acceso por posición (como en una lista)
print(poblacion.iloc[0])
````

La diferencia con las listas y los diccionarios aparece en las operaciones: una expresión aritmética se aplica a **todos los valores** de la Series a la vez, y el resultado es otra Series con el mismo índice. Esta es la vectorización heredada de NumPy:

````python
# Población en miles de habitantes: la división se aplica a cada valor
poblacion / 1000
````

Compárese con la serie *Fundamentos de Python*, donde el mismo cálculo requería un ciclo `for` que recorriera una lista. Las Series tienen además numerosos métodos estadísticos:

````python
# Estadísticas de la Series
print(poblacion.sum())
print(poblacion.mean())
print(poblacion.max())

# idxmax() retorna la ETIQUETA del valor máximo
print(poblacion.idxmax())
````

## DataFrames

Un **DataFrame** es una tabla: un conjunto de Series que comparten el mismo índice, una por columna. Puede crearse directamente a partir de una lista de diccionarios —la estructura con que cerró la serie anterior— en la que cada diccionario se convierte en una fila y cada clave, en una columna:

````python
# Provincias de Costa Rica: población y área
# (INEC, Estimación de Población y Vivienda 2022)
provincias = [
    {"nombre": "San José", "poblacion": 1601167, "area_km2": 4969.80},
    {"nombre": "Alajuela", "poblacion": 1035464, "area_km2": 9772.12},
    {"nombre": "Cartago", "poblacion": 545092, "area_km2": 3093.24},
    {"nombre": "Heredia", "poblacion": 479117, "area_km2": 2663.30},
    {"nombre": "Guanacaste", "poblacion": 412808, "area_km2": 10196.29},
    {"nombre": "Puntarenas", "poblacion": 500166, "area_km2": 11275.00},
    {"nombre": "Limón", "poblacion": 470383, "area_km2": 9176.88}
]

df_provincias = pd.DataFrame(provincias)

df_provincias
````

Cuando la última expresión de una celda es un DataFrame, Jupyter lo presenta como una tabla con formato. Sus componentes son el **índice** (a la izquierda, aquí las posiciones 0 a 6), los nombres de las **columnas** (arriba) y los **valores**. Tres atributos los describen:

````python
# Dimensiones: (filas, columnas)
print(df_provincias.shape)

# Nombres de las columnas
print(df_provincias.columns)

# Tipos de datos de cada columna
print(df_provincias.dtypes)
````

Cada columna de un DataFrame es una Series, que se obtiene con el nombre de la columna entre corchetes. Por lo tanto, las operaciones vectorizadas recién vistas funcionan entre columnas, y la densidad de población de las siete provincias se calcula en una sola expresión, sin el ciclo `for` de la serie anterior:

````python
# Una columna es una Series
df_provincias["poblacion"]
````

````python
# Densidad de población: operación entre dos columnas completas
df_provincias["poblacion"] / df_provincias["area_km2"]
````

El resultado es una Series nueva; en la sección de columnas nuevas se verá cómo agregarla al DataFrame.

## Selección de columnas y filas

### Columnas

Ya se vio que una columna se obtiene con su nombre entre corchetes. Para seleccionar **varias columnas** se pasa una **lista** de nombres, es decir, corchetes dentro de corchetes. La distinción es importante: con un nombre se obtiene una Series y con una lista, un DataFrame (aunque la lista tenga un solo nombre):

````python
# Una columna: Series
cantones["canton"].head()
````

````python
# Varias columnas: DataFrame (note los corchetes dobles)
cantones[["canton", "poblacion", "densidad"]].head()
````

### Filas

Las filas se seleccionan con los mismos accesores de las Series: `loc[]` por etiqueta del índice e `iloc[]` por posición. Como el índice de `cantones` es el de posiciones que asignó `read_csv()`, ambos coinciden en este caso, con una diferencia: en los rangos, `loc[]` **incluye** el límite final e `iloc[]` lo excluye, como el rebanado de las listas:

````python
# Primera fila: una Series cuyo índice son los nombres de las columnas
cantones.iloc[0]
````

````python
# Filas en las posiciones 0, 1 y 2 (la 3 se excluye)
cantones.iloc[0:3]
````

````python
# Filas con etiquetas 0 a 2 (inclusive) y solo algunas columnas
cantones.loc[0:2, ["canton", "poblacion"]]
````

`loc[]` acepta filas y columnas a la vez, separadas por una coma, y es la forma general de seleccionar en pandas. Se usa con mucha más frecuencia que `iloc[]`, porque las etiquetas son más significativas que las posiciones, y es también la forma de seleccionar filas por condición, como se ve a continuación.

### Filtrado por condiciones

La selección más útil es la **condicional**: quedarse con las filas que cumplen un criterio. Funciona en dos pasos. Primero, una comparación con una columna produce una Series de valores booleanos, uno por fila:

````python
# Una comparación vectorizada produce una Series booleana
cantones["poblacion"] > 100000
````

Segundo, esa Series booleana entre corchetes selecciona las filas donde el valor es `True`. Es lo que se conoce como **máscara booleana** (*boolean mask*):

````python
# Cantones con más de 100 000 habitantes
cantones[cantones["poblacion"] > 100000]
````

Las condiciones se combinan con los operadores `&` (y), `|` (o) y `~` (no), **cada condición entre paréntesis**. No sirven `and`, `or` y `not`, que operan sobre valores individuales y no sobre Series (la clínica de errores muestra qué ocurre si se usan):

````python
# Cantones de Guanacaste con menos de 30 habitantes por km2
cantones[(cantones["provincia"] == "Guanacaste") & (cantones["densidad"] < 30)]
````

````python
# Cantones de Limón o de Puntarenas con más de 50 000 habitantes,
# solo con algunas columnas
cantones.loc[
    (cantones["provincia"].isin(["Limón", "Puntarenas"])) & (cantones["poblacion"] > 50000),
    ["provincia", "canton", "poblacion"]
]
````

El método `isin()` verifica la pertenencia a una lista de valores y reemplaza una cadena de condiciones unidas con `|`. Por último, la cantidad de filas que cumplen una condición se obtiene con `len()` del resultado o, más directamente, sumando la Series booleana (`True` cuenta como 1):

````python
# ¿Cuántos cantones tienen más de 1000 habitantes por km2?
(cantones["densidad"] > 1000).sum()
````

## Ordenamiento

El método `sort_values()` ordena las filas según una o más columnas. Igual que la función `sorted()` de la serie anterior, **retorna un DataFrame nuevo** y deja el original intacto:

````python
# Los cinco cantones más poblados
cantones.sort_values("poblacion", ascending=False).head()
````

````python
# Ordenamiento por provincia y, dentro de cada provincia, por densidad descendente
cantones.sort_values(["provincia", "densidad"], ascending=[True, False]).head(10)
````

Note el **encadenamiento de métodos**: `sort_values()` retorna un DataFrame, sobre el que se invoca `head()`. Este estilo, en el que cada método transforma el resultado del anterior, es característico de pandas y se leerá de izquierda a derecha como una secuencia de pasos.

## Columnas nuevas

Una columna nueva se crea asignando una Series —típicamente, el resultado de una operación entre columnas existentes— a un nombre de columna que no existe. A diferencia de `sort_values()`, la asignación **modifica el DataFrame** en el lugar:

````python
# Porcentaje de mujeres en la población de cada cantón
cantones["porcentaje_mujeres"] = cantones["mujeres"] / cantones["poblacion"] * 100

cantones[["canton", "poblacion", "mujeres", "porcentaje_mujeres"]].head()
````

Las columnas derivadas sirven también para **verificar** los datos. La densidad que publica el INEC debería coincidir con la población dividida entre el área; el siguiente código la recalcula y comprueba, fila por fila y sin ciclos, que ambas coinciden:

````python
# Densidad recalculada, redondeada a un decimal como la del INEC
cantones["densidad_calculada"] = round(cantones["poblacion"] / cantones["area_km2"], 1)

# ¿Coinciden todas las filas? all() retorna True solo si todos los valores son True
(cantones["densidad_calculada"] == cantones["densidad"]).all()
````

Una columna se elimina con `drop()`, que, como `sort_values()`, retorna un DataFrame nuevo, por lo que el resultado debe asignarse:

````python
# Eliminación de la columna de verificación
cantones = cantones.drop(columns="densidad_calculada")

cantones.columns
````

### Exportación

El camino inverso a `read_csv()` es el método `to_csv()`, que guarda el DataFrame en un archivo. El argumento `index=False` evita que el índice de posiciones se escriba como una columna más:

````python
# Cantones de Heredia ordenados por densidad, guardados en un archivo CSV
heredia = cantones[cantones["provincia"] == "Heredia"].sort_values("densidad", ascending=False)
heredia.to_csv("cantones-heredia.csv", index=False)

# Verificación: se vuelve a leer el archivo recién escrito
pd.read_csv("cantones-heredia.csv").head(3)
````

## Clínica de errores

Los errores típicos del trabajo con DataFrames, provocados a propósito para aprender a reconocerlos:

````python
# KeyError: la columna no existe
# (aquí, por una errata en el nombre; la última línea del mensaje muestra el nombre buscado)
cantones["poblacón"]
````

````python
# ValueError: "The truth value of a Series is ambiguous"
# (por combinar condiciones con and en vez de &)
cantones[(cantones["provincia"] == "Limón") and (cantones["poblacion"] > 50000)]
````

````python
# AttributeError: el DataFrame no tiene ese método
# (aquí, por una errata en el nombre del método)
cantones.haed()
````

El segundo error merece atención porque su mensaje es poco intuitivo. `and` intenta decidir si una Series completa es verdadera o falsa, lo que no tiene sentido cuando contiene muchos valores; de ahí lo de "ambiguo". La propia última línea del mensaje sugiere la solución, aunque de manera críptica: usar los operadores vectorizados. Es un buen candidato para pedir una explicación a un asistente de IA con la estructura de prompt de la [serie anterior](https://gf0657-programacionsig.github.io/2026-ii/estructuras-datos-apis/) (rol, contexto, tarea y formato), como propone uno de los ejercicios.

## Resumen

- **pandas** es la biblioteca de Python para datos tabulares. Sus operaciones son **vectorizadas**: se aplican a columnas completas, sin ciclos.
- Una **Series** es una columna con **índice**; un **DataFrame** es una tabla de Series con el mismo índice. Un DataFrame se crea desde una **lista de diccionarios** o se carga desde un archivo **CSV** con `pd.read_csv()`, incluso desde una URL.
- Exploración inicial: `head()`, `shape`, `info()`, `describe()` y `value_counts()`.
- Selección: `df["columna"]` (Series), `df[["c1", "c2"]]` (DataFrame), `loc[]` por etiquetas e `iloc[]` por posiciones.
- Filtrado con **máscaras booleanas**: `df[condición]`, con `&`, `|`, `~` y paréntesis para combinar condiciones, e `isin()` para listas de valores.
- `sort_values()` y `drop()` retornan un DataFrame nuevo; la asignación `df["nueva"] = ...` modifica el DataFrame. `to_csv()` exporta.
- Errores típicos: `KeyError` (columna inexistente), `ValueError` por combinar condiciones con `and`/`or` y `AttributeError` (método inexistente).
